# **Exploratory Data Analysis (EDA)**

# Region-wise average unemployment rates

```python
df.head(5)
```

```python
region_avg=df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False)
```

```python
region_avg
```

```python
plt.figure(figsize=(8,6))
sns.barplot(y=region_avg.index
            ,x=region_avg.values
            ,palette="viridis"
            , hue=region_avg
            , legend=False
            )
plt.title('Region-wise average unemployment rates')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('Region')
plt.show()

```

# Markdown Observation

### Observation:
The average unemployment rate differs significantly across regions. Some states consistently report higher unemployment than others, indicating regional disparities in employment opportunities

# Month-wise Trend

```python
# Create Month Column
df["Month"]=df["Date"].dt.month_name()
```

```python
df.head()
```

```python
# Month Order
month_ord=["January","February","March","April","May","June",
    "July","August","September","October","November","December"]
```

```python
month_trend=df.groupby("Month")["Estimated Unemployment Rate (%)"].mean().reindex(month_ord)
```

```python
month_trend
```

```python
sns.lineplot(x=month_trend.index,
             y=month_trend.values,
             color="r",
             marker="o",
             linestyle="--")
plt.title('Month-wise Average Unemployment Rate')
plt.xticks(rotation=40)
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')
plt.grid(True)
plt.show()
```

# Markdown Observation

### Observation:
Monthly unemployment rates fluctuate over time, with a noticeable increase during the COVID-19 period followed by gradual recovery.

# Time-Series Line Chart

```python
df
```

```python

states=["Delhi", "Maharashtra", "West Bengal"]

```

```python
plt.figure(figsize=(12,6))
plt.grid(True)
for state in states :
  temp=df[df["Region"]==state]

  plt.plot(temp["Date"],
           temp["Estimated Unemployment Rate (%)"],
           marker="o",
           linestyle="--",
           linewidth=1,
           label=state)

  plt.ylabel("Unemployment Rate Over Time")
  plt.xlabel("Timestamp")
  plt.title("Time-Series Line Chart")
  plt.legend(title="States")
  plt.xticks(rotation=40)


plt.show()
```

# Markdown Observation

### Observation:
All three states experienced fluctuations in unemployment over time. A sharp increase is visible around the COVID-19 period, especially in Delhi.

# Pre-COVID vs Post-COVID Comparison

```python
# Split DataSet Based on the Covid Timeline
preCovid=df[df["Date"]<"2020-03-01"]
postCovid=df[df["Date"]>"2020-03-01"]
```

```python
preCovid.sample(5)
```

```python
postCovid.sample(5)
```

```python
comparison=pd.DataFrame(
  {
      "Period":["Pre Covid","Post Covid"],
      "Avg_Unemployment_Rate":[
          
          preCovid["Estimated Unemployment Rate (%)"].mean(),
          postCovid["Estimated Unemployment Rate (%)"].mean()
      ]
  }
)
```

```python
comparison
```

## Bar Chart

```python
plt.figure(figsize=(8,6))
sns.barplot(x=comparison.Period,y=comparison.Avg_Unemployment_Rate,palette="BrBG",hue=comparison.Period)
plt.ylabel("Average Unemployment Rate (%)")
plt.xlabel("Period")
plt.title("Pre-COVID vs Post-COVID Average Unemployment Rate")
plt.xticks(rotation=35)
plt.show()
```

# Markdown Observation

### Observation:
The average unemployment rate increased after March 2020, reflecting the significant economic impact of the COVID-19 pandemic on employment in India.
